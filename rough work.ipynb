```python
vowels = ('a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U')

letter = input("Enter a single letter: ")

if len(letter) == 1 and letter.isalpha():
    if letter in vowels:
        print(f"{letter} is a Vowel")
    else:
        print(f"{letter} is a Consonant")
else:
    print("Please enter valid alphabet")
```
