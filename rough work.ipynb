```python
def calculate_rectangle(lenght, width):
    
    def calculate_area():
        return(lenght * width)
        
    def calculate_perimeter():
        return(2 * (lenght + width))
    
    print(f"Area: {calculate_area()}")
    print(f"Perimeter: {calculate_perimeter()}")
   
calculate_rectangle(3, 5)
    
```

```python
def validate_user(correct_username, correct_password):
    # Outer function stores the valid credentials
    
    def check_credentials(username, password):
        # Inner function validates the input against the stored credentials
        if username == correct_username and password == correct_password:
            return "Login Successful! Access granted."
        else:
            return "Login Failed! Invalid username or password."
            
    return check_credentials

# --- Example Usage ---
# Setup valid credentials
login_checker = validate_user("admin123", "secret_password")

# Test with correct details
print(login_checker("admin123", "secret_password"))

# Test with incorrect details
print(login_checker("admin123", "wrong_password"))
```

```python
def shopping_bill(price, quantity, discount=0):
    
    def calculate_total():
        print(price * quantity)
        
    def apply_discount(total):
        final_amount = total - discount
        return max(0, final_amount)

    total = calculate_total()
    final_bill = apply_discount(total)
    
    return final_bill

        
bill_without_discount = shopping_bill(100, 6, discount=0)
print(f'Final Bill = {bill_without_discount} (no discount)')

bill_with_discount = shopping_bill(100, 6, discount=100)
print(f'Final Bill = {bill_with_discount} (with discount)')
```
