```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print(f'name : {self.name}')
        print(f'number : {self.number}')
        print(f'address : {self.address}')
        print(f'salary : {self.salary}')
        print(f'post : {self.post}')
        print(f'year_of_joining : {self.year_of_joining}')
        print(f'education : {self.education}')
        print(f'year_of_experience : {self.year_of_experience}')

    def update_year_of_experience(self, additional_years):
        self.year_of_experience += additional_years
        print(f'Updated Experience of {self.name} : {self.year_of_experience} years')

    def increase_salary(self,percentage):
        increament_amount = self.salary * (percentage/100)
        self.salary += increament_amount
        print(f'Increase salary of {self.name} by {percentage}% = {self.salary}')
```
