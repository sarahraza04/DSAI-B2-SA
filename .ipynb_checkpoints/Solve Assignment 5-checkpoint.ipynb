# Q1

```python
def count_vowels(word):
    vowels =  'aeiouAEIOU'
    count = 0
    for i in word:
        if i in vowels:
            count += 1
    return(f'Output : {count} vowels')
        

count_vowels('Welcome')
```

# Q2

```python
def info(**kargs):
    for key, value in kargs.items():
        print(f'{key} : {value}')
info(name='John Doe', age=15, Grade='10th', School='Greenwood High') 
```

# Q3

```python
def display_prices(product_name, **kargs):
    print(f"Product: {product_name}")
    for key, value in kargs.items():
        print(f'{key} : {value}')
display_prices('Laptop', Amazon = '$800', eBay = '$750' , Walmart = '$780')
```

# Q6

```python
def calculate_rectangle(lenght, width):
    
    def calculate_area():
        return(lenght * width)
        
    def calculate_perimeter():
        return(2 * (lenght + width))
    
    print(f"Area: {calculate_area()}")
    print(f"Perimeter: {calculate_perimeter()}")
   
calculate_rectangle(3, 5)
```

# Q8

```python
def total_bill(num_of_nights):
    def cost():
        return num_of_nights * 5000
    def tax():
        return cost()*10/100
    print(f'Room Cost: {cost()}')
    print(f'Tax: {tax()}')
    bill = cost()+tax()
    print(f'Total Bill: {bill}')
total_bill(3)
```

# Q9

```python
def login_validation(username, password):
    def check_credential(input_username, input_password):
        if username == input_username and password == input_password:
            print("Access Granted: Login Successful!")
        else:
            print("Access Denied: Invalid username or password.")
    return(check_credential)
login_checker = login_validation("admin123", "secret_password")

login_checker("admin123", "secret_password")

login_checker("admin123", "correct_password")
```

# Q10

```python
def shopping_bill(price, quantity, discount=0):
    def calculate_total():
        return price * quantity
    def apply_discount(total):
        final_amount = total - discount
        return max(0, final_amount)

    total = calculate_total()
    final_bill = apply_discount(total)
    
    return final_bill

bill_no_discount = shopping_bill(100, 6, discount=0)
print(f"Final Bill = {bill_no_discount} PKR (no discount)")

bill_with_discount = shopping_bill(100, 6, discount=100)
print(f"Final Bill = {bill_with_discount} PKR (with discount)")
```
