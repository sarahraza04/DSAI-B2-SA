# Q2

```python
age = int(input("age group classification"))

if age >= 0 and age <= 12:
    print("Child")
elif age >= 13 and age <= 19:
    print("Teenager")
elif age >= 20 and age <= 64:
    print("Adult")
else:
    print("Senior")
```

```python
vowels = ('a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U')

letter = input("Enter a single letter: ")

if letter.isalpha():
    if letter in vowel:
        print("Vowel")
    else:
        print("Consonant")
else:
    print("Please enter valid alphabet")
    
```
