```python
# Initial product catalog using a nested dictionary
products = {
    "P1": {"name": "Shoes", "price": 2500, "stock": 5},
    "P2": {"name": "Shirt", "price": 1500, "stock": 10},
    "P3": {"name": "Bag", "price": 3000, "stock": 4},
    "P4": {"name": "Watch", "price": 4000, "stock": 3},
    "P5": {"name": "Headphones", "price": 2000, "stock": 6},
}

# Shopping cart: maps product_id -> quantity
cart = {}


def view_products():
    print("\n--- Available Products ---")
    print(f"{'ID':<5} {'Product':<15} {'Price (Rs.)':<12} {'Stock':<6}")
    print("-" * 42)
    for pid, info in products.items():
        print(f"{pid:<5} {info['name']:<15} {info['price']:<12,} {info['stock']:<6}")
    print()


def add_to_cart():
    view_products()
    pid = input("Enter Product ID (e.g., P1): ").strip().upper()

    if pid not in products:
        print("Invalid Product ID. Please try again.\n")
        return

    try:
        qty = int(input(f"Enter quantity for {products[pid]['name']}: "))
    except ValueError:
        print("Please enter a valid whole number for quantity.\n")
        return

    if qty <= 0:
        print("Quantity must be greater than zero.\n")
        return

    current_in_cart = cart.get(pid, 0)
    total_requested = current_in_cart + qty
    available_stock = products[pid]["stock"]

    if total_requested > available_stock:
        print(
            f"Sorry! Only {available_stock} unit(s) of {products[pid]['name']} "
            f"available in stock (you currently have {current_in_cart} in your cart).\n"
        )
        return

    cart[pid] = total_requested
    print(f"Added {qty} x {products[pid]['name']} to your cart. (Total in cart: {cart[pid]})\n")


def view_cart():
    if not cart:
        print("\nYour cart is currently empty.\n")
        return 0

    print("\n--- Your Shopping Cart ---")
    print(f"{'ID':<5} {'Product':<15} {'Price':<10} {'Qty':<6} {'Subtotal':<12}")
    print("-" * 52)

    cart_total = 0
    for pid, qty in cart.items():
        name = products[pid]["name"]
        price = products[pid]["price"]
        subtotal = price * qty
        cart_total += subtotal
        print(f"{pid:<5} {name:<15} Rs.{price:<7,} {qty:<6} Rs.{subtotal:<9,}")

    print("-" * 52)
    print(f"{'Cart Total:':<40} Rs.{cart_total:,}\n")
    return cart_total


def checkout():
    if not cart:
        print("\nYour cart is empty. Add items before checking out.\n")
        return

    cart_total = view_cart()

    discount = 0
    if cart_total >= 5000:
        discount = int(cart_total * 0.10)

    final_bill = cart_total - discount

    print("=== Checkout Summary ===")
    print(f"Cart Total:   Rs.{cart_total:,}")
    if discount > 0:
        print(f"Discount (10%): -Rs.{discount:,}")
    else:
        print("Discount:     Rs.0 (10% off applies on orders of Rs. 5,000 or more)")
    print(f"Final Bill:   Rs.{final_bill:,}")
    print("========================")

    # Deduct stock and clear cart
    for pid, qty in cart.items():
        products[pid]["stock"] -= qty

    cart.clear()
    print("Thank you for your purchase! Stock has been updated and your cart is now empty.\n")


def main():
    while True:
        print("=== Online Shopping Cart Menu ===")
        print("1. View Products")
        print("2. Add to Cart")
        print("3. View Cart")
        print("4. Checkout")
        print("5. Exit")

        choice = input("Choose an option (1-5): ").strip()

        if choice == "1":
            view_products()
        elif choice == "2":
            add_to_cart()
        elif choice == "3":
            view_cart()
        elif choice == "4":
            checkout()
        elif choice == "5":
            print("Exiting program. Goodbye!")
            break
        else:
            print("Invalid choice. Please select a number from 1 to 5.\n")


main()
```
