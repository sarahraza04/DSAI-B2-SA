```python
Products = {
    'P1': {'Product': 'Shoes', 'Price': 2500, 'Available_Stock': 5},
    'P2': {'Product': 'Shirt', 'Price': 1500, 'Available_Stock': 10},
    'P3': {'Product': 'Bag', 'Price': 3000, 'Available_Stock': 4},
    'P4': {'Product': 'Watch', 'Price': 4000, 'Available_Stock': 3},
    'P5': {'Product': 'Headphones', 'Price': 2000, 'Available_Stock': 6},
}

cart = {}

while True:
    print("\n===== ONLINE SHOPPING CART =====")
    print("1. View Products")
    print("2. Add to Cart")
    print("3. View Cart")
    print("4. Checkout")
    print("5. Exit")
    choice = input("Enter your choice (1-5): ").strip()

    if choice == "1":
        print("\n" + "-" * 50)
        print(f"{'ID':<6}{'Product':<15}{'Price':<12}{'Available_Stock':<15}")
        print("-" * 50)
        for pid, info in Products.items():
            print(f"{pid:<6}{info['Product']:<15}Rs. {info['Price']:<8}{info['Available_Stock']:<15}")
        print("-" * 50)

    elif choice == "2":
        pid = input("Enter Product ID between P1-P5: ").strip().upper()
        if pid not in Products:
            print("Invalid Product ID. Enter a correct Product ID.")
        else:
            try:
                quantity = int(input("Enter quantity: "))
            except ValueError:
                print("Please enter a valid whole number.")
                continue

            if quantity <= 0:
                print("Enter a number greater than 0.")
            else:
                already_in_cart = cart.get(pid, 0)
                total_requested = already_in_cart + quantity
                available = Products[pid]["Available_Stock"]
                if total_requested > available:
                    print(
                        f"Out of Stock! Only {available} {Products[pid]['Product']} available "
                        f"({already_in_cart} already in your cart)."
                    )
                else:
                    cart[pid] = total_requested
                    print(
                        f"Added {quantity} x {Products[pid]['Product']} to cart. "
                        f"(Total in cart: {total_requested})"
                    )

    elif choice == "3":
        if len(cart) == 0:
            print("Your cart is empty.")
        else:
            cart_total = 0
            print("\n" + "-" * 60)
            print(f"{'Product':<15}{'Qty':<8}{'Price':<12}{'Subtotal':<12}")
            print("-" * 60)
            for pid, quantity in cart.items():
                name = Products[pid]["Product"]
                price = Products[pid]["Price"]
                subtotal = price * quantity
                cart_total += subtotal
                print(f"{name:<15}{quantity:<8}Rs. {price:<8}Rs. {subtotal:<8}")
            print("-" * 60)
            print(f"Cart Total: Rs. {cart_total}")

    elif choice == "4":
        if len(cart) == 0:
            print("Your cart is empty.")
        else:
            cart_total = sum(Products[pid]["Price"] * qty for pid, qty in cart.items())
            discount = cart_total * 0.10 if cart_total >= 5000 else 0
            final_bill = cart_total - discount

            print("\n" + "=" * 40)
            print("               FINAL BILL")
            print("=" * 40)
            for pid, qty in cart.items():
                line_total = Products[pid]["Price"] * qty
                print(f"{Products[pid]['Product']} x {qty} = Rs. {line_total}")
            print("-" * 40)
            print(f"Cart Total : Rs. {cart_total:.2f}")
            print(f"Discount   : Rs. {discount:.2f}")
            print(f"Final Bill : Rs. {final_bill:.2f}")
            print("=" * 40)

            for pid, qty in cart.items():
                Products[pid]["Available_Stock"] -= qty
            cart.clear()
            print("Checkout complete. Thank you for shopping with us!")

    elif choice == "5":
        print("Thank you for visiting. Goodbye!")
        break

    else:
        print("Invalid choice. Please select a number from 1 to 5.")
```
